### Filling the Review Rating Column Value with median

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

### Modify the table columns name (as lower, underscore and rename)

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', "_")
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

### Create a custom column (i.e., age group)

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q= 4, labels = labels)
```

```python
df[['age', 'age_group']].head()
```

### Purchase Ferequency Days

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequenncy_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
